# 🚀 PhantomX V2 & V3 Dual Engine - Master Google Colab 1-Click Runtime

> **Architecture**: Surgical 💉 | Aviation ✈️ | Military 🪖
> **Single 1-Click Execution**: Cell 1 mounts Drive, activates 24/7 Keep-Alive JS, unzips engine code, and launches Live Mainnet Orchestrator (`DRY_RUN=false`).
> **Telegram Integration**: Direct live 10-min and 30-min telemetry status sent to Telegram.

```python
# ==============================================================================
# CELL 1: 1-Click Mount Drive, Keep-Alive, Install Packages & Launch Orchestrator
# ==============================================================================
import os
import sys
import time
import shutil
import json
import zipfile
import subprocess
import threading
from datetime import datetime

# 0. Activate Colab 24/7 Keep-Alive JS Loop
try:
    from IPython.display import display, Javascript
    js_code = """
    function KeepAlive() {
        console.log("Keeping Colab Alive 24/7...");
        var btn = document.querySelector("colab-connect-button") || document.querySelector("#connect");
        if (btn) { btn.click(); }
    }
    setInterval(KeepAlive, 60000);
    """
    display(Javascript(js_code))
    print("✅ Colab 24/7 Keep-Alive Loop Activated in Cell 1!")
except Exception as e:
    print(f"ℹ️ Keep-Alive note: {e}")

# 1. Mount Google Drive
if 'google.colab' in sys.modules:
    from google.colab import drive
    drive.mount('/content/drive', force_remount=False)

# 2. Locate Drive Source (Zip or Folder)
POSSIBLE_ZIP_PATHS = [
    '/content/drive/MyDrive/v2_v3_engine_colab_ready.zip',
    '/content/drive/MyDrive/flash loan ghost hunter/v2_v3_engine_colab_ready.zip'
]
POSSIBLE_FOLDER_PATHS = [
    '/content/drive/MyDrive/v2_v3_engine',
    '/content/drive/MyDrive/flash loan ghost hunter/v2_v3_engine'
]

VM_WORKSPACE = '/content/v2_v3_engine'
drive_backup_dir = '/content/drive/MyDrive/v2_v3_engine_logs'
os.makedirs(drive_backup_dir, exist_ok=True)

found_source = False
# Check ZIP first
for zp in POSSIBLE_ZIP_PATHS:
    if os.path.exists(zp):
        print(f'📦 Found Colab-Ready ZIP at: {zp}')
        print(f'🚀 Unzipping to Fast VM Disk: {VM_WORKSPACE}...')
        if os.path.exists(VM_WORKSPACE):
            shutil.rmtree(VM_WORKSPACE, ignore_errors=True)
        os.makedirs(VM_WORKSPACE, exist_ok=True)
        with zipfile.ZipFile(zp, 'r') as zf:
            zf.extractall(VM_WORKSPACE)
        found_source = True
        break

# Fallback to folder
if not found_source:
    for fp in POSSIBLE_FOLDER_PATHS:
        if os.path.exists(fp):
            print(f'📁 Found v2_v3_engine Folder at: {fp}')
            print(f'🚀 Copying folder to Fast VM Disk: {VM_WORKSPACE}...')
            if os.path.exists(VM_WORKSPACE):
                shutil.rmtree(VM_WORKSPACE, ignore_errors=True)
            shutil.copytree(fp, VM_WORKSPACE, ignore=shutil.ignore_patterns('*.pyc', '__pycache__', '.pytest_cache', '*.jsonl'))
            found_source = True
            break

if not found_source:
    print('⚠️ Source not found on Drive. Using current working directory.')
    VM_WORKSPACE = os.getcwd()

# Handle nested folder extraction if zip unzipped into a subfolder
nested_dir = os.path.join(VM_WORKSPACE, 'v2_v3_engine')
if os.path.exists(os.path.join(nested_dir, 'common', 'master_background_orchestrator.py')):
    VM_WORKSPACE = nested_dir

os.chdir(VM_WORKSPACE)
if VM_WORKSPACE not in sys.path:
    sys.path.insert(0, VM_WORKSPACE)
print(f'✅ Active Fast VM Workspace: {VM_WORKSPACE}')

# 3. Install Dependencies
print('📦 Auditing & Installing Dependencies...')
reqs = ['web3', 'scikit-learn', 'numpy', 'pandas', 'requests', 'psutil', 'python-dotenv']
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q'] + reqs)
print('✅ All Dependencies Active!')

# 4. 5-Minute Google Drive Auto-Sync Loop
def drive_sync_loop():
    while True:
        try:
            for root, dirs, files in os.walk(VM_WORKSPACE):
                for file in files:
                    if file.endswith('.log') or file.endswith('.jsonl') or file.endswith('.md') or file.endswith('.json'):
                        src = os.path.join(root, file)
                        dst = os.path.join(drive_backup_dir, file)
                        try:
                            shutil.copy2(src, dst)
                        except Exception:
                            pass
            print(f'💾 [Drive Auto-Sync] Telemetry logs backed up to Drive ({datetime.now().strftime("%H:%M:%S")})')
        except Exception as e:
            print(f'⚠️ Drive Sync Warning: {e}')
        time.sleep(300)

sync_thread = threading.Thread(target=drive_sync_loop, daemon=True)
sync_thread.start()
print('🔄 Drive Auto-Sync active (Syncs every 5 mins).')

# 5. Launch Master Background Orchestrator in LIVE Mainnet Mode (DRY_RUN=false)
orchestrator_script = os.path.join(VM_WORKSPACE, 'common', 'master_background_orchestrator.py')
print(f'⚡ Spawning Master Background Orchestrator (LIVE MAINNET MODE): {orchestrator_script}')
p = subprocess.Popen([sys.executable, orchestrator_script, '--mode', 'live'], cwd=os.path.join(VM_WORKSPACE, 'common'))

print('==============================================================================')
print('🎉 V2 & V3 DUAL ENGINE RUNNING LIVE ON GOOGLE COLAB (DRY_RUN=false)!')
print('==============================================================================')
print('📡 Live Telemetry & Telegram Alerts are ACTIVE.')
print('💾 5-Minute Google Drive Auto-Sync is ACTIVE.')
print('==============================================================================')

p.wait()

```
